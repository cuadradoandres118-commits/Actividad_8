````python
# Actividad 8. Códigos SQL de AgroDatos

## 1. Ejercicio 1. Modificar la base de datos

### 1.1. Agregar el tipo de riego a la tabla `lote`

```sql
ALTER TABLE lote
ADD COLUMN tipo_riego ENUM(
    'goteo',
    'aspersión',
    'gravedad',
    'secano'
) NOT NULL DEFAULT 'secano';
```

### 1.2. Crear la tabla `operario`

```sql
CREATE TABLE operario (
    id_operario INT AUTO_INCREMENT PRIMARY KEY,
    documento VARCHAR(20) NOT NULL UNIQUE,
    nombre_completo VARCHAR(100) NOT NULL,
    telefono VARCHAR(20),
    certificado_aplicador BOOLEAN NOT NULL DEFAULT FALSE
);
```

### 1.3. Insertar los operarios

```sql
INSERT INTO operario (
    documento,
    nombre_completo,
    telefono,
    certificado_aplicador
)
VALUES
('1012', 'Jorge Patiño', '3101112233', TRUE),
('2045', 'Luz Marina Ortiz', '3125556677', TRUE),
('3090', 'Ramiro Suárez', '3009998877', FALSE);
```

## 2. Ejercicio 2. Productos y aplicaciones

### 2.1. Insertar los productos faltantes

```sql
INSERT INTO producto (
    nombre_comercial,
    tipo,
    ingrediente_activo,
    categoria_tox,
    registro_ica,
    dosis_ha,
    unidad,
    carencia_dias,
    reingreso_horas,
    precio_unidad
)
VALUES
(
    'BioLarva WP',
    'bioinsumo',
    'Bacillus thuringiensis var. kurstaki',
    'IV',
    'ICA-DEMO-2001',
    0.5,
    'kg',
    0,
    4,
    95000
),
(
    'RoyaStop 250 EC',
    'fungicida',
    'Triazol 250 g/L',
    'III',
    'ICA-DEMO-3001',
    0.4,
    'L',
    30,
    24,
    160000
),
(
    'TizónFin 80 WP',
    'fungicida',
    'Ditiocarbamato 800 g/kg',
    'III',
    'ICA-DEMO-3002',
    2.5,
    'kg',
    14,
    24,
    38000
);
```

### 2.2. Registrar la aplicación en el lote POR-L01

```sql
INSERT INTO aplicacion (
    id_lote,
    id_producto,
    fecha,
    cantidad,
    responsable
)
SELECT
    l.id_lote,
    p.id_producto,
    '2026-10-14',
    p.dosis_ha * l.area_ha,
    'Estudiante'
FROM lote AS l
CROSS JOIN producto AS p
WHERE l.codigo = 'POR-L01'
  AND p.nombre_comercial = 'RoyaStop 250 EC';
```

### 2.3. Calcular la fecha mínima de cosecha

```sql
SELECT
    a.id_aplicacion,
    a.fecha AS fecha_aplicacion,
    l.codigo AS lote,
    p.nombre_comercial AS producto,
    p.carencia_dias,
    DATE_ADD(
        a.fecha,
        INTERVAL p.carencia_dias DAY
    ) AS fecha_minima_cosecha
FROM aplicacion AS a
INNER JOIN lote AS l
    ON a.id_lote = l.id_lote
INNER JOIN producto AS p
    ON a.id_producto = p.id_producto
WHERE l.codigo = 'POR-L01'
  AND p.nombre_comercial = 'RoyaStop 250 EC';
```

## 3. Ejercicio 3. Consultas SQL

### 3.1. Área total por cultivo

```sql
SELECT
    cultivo,
    SUM(area_ha) AS area_total_ha
FROM lote
GROUP BY cultivo
ORDER BY area_total_ha DESC;
```

### 3.2. Fincas con más de 5.0 hectáreas sembradas

```sql
SELECT
    f.id_finca,
    f.nombre AS finca,
    SUM(l.area_ha) AS area_total_ha
FROM finca AS f
INNER JOIN lote AS l
    ON f.id_finca = l.id_finca
GROUP BY f.id_finca, f.nombre
HAVING SUM(l.area_ha) > 5.0
ORDER BY area_total_ha DESC;
```

### 3.3. Productos que nunca se han aplicado

```sql
SELECT
    p.id_producto,
    p.nombre_comercial,
    p.tipo
FROM producto AS p
LEFT JOIN aplicacion AS a
    ON p.id_producto = a.id_producto
WHERE a.id_aplicacion IS NULL;
```

### 3.4. Responsable que aplicó más kilogramos de fertilizante

```sql
SELECT
    a.responsable,
    SUM(a.cantidad) AS kilogramos_aplicados
FROM aplicacion AS a
INNER JOIN producto AS p
    ON a.id_producto = p.id_producto
WHERE p.tipo = 'fertilizante'
  AND p.unidad = 'kg'
GROUP BY a.responsable
ORDER BY kilogramos_aplicados DESC
LIMIT 1;
```

## 4. Ejercicio 4. Normalización

### 4.1. Crear la tabla `aplicacion_n`

```sql
CREATE TABLE aplicacion_n (
    id_aplicacion_n INT AUTO_INCREMENT PRIMARY KEY,
    fecha DATE NOT NULL,
    id_lote INT NOT NULL,
    id_producto INT NOT NULL,
    cantidad DECIMAL(10,2) NOT NULL,
    documento_operario VARCHAR(20) NOT NULL,
    observaciones VARCHAR(255),

    CONSTRAINT fk_aplicacion_n_lote
        FOREIGN KEY (id_lote)
        REFERENCES lote(id_lote),

    CONSTRAINT fk_aplicacion_n_producto
        FOREIGN KEY (id_producto)
        REFERENCES producto(id_producto),

    CONSTRAINT fk_aplicacion_n_operario
        FOREIGN KEY (documento_operario)
        REFERENCES operario_n(documento)
);
```

### 4.2. Migrar los datos de `planilla_campo`

```sql
INSERT INTO aplicacion_n (
    fecha,
    id_lote,
    id_producto,
    cantidad,
    documento_operario,
    observaciones
)
SELECT
    pc.fecha,
    l.id_lote,
    p.id_producto,
    pc.cantidad,
    pc.documento_operario,
    pc.observaciones
FROM planilla_campo AS pc
INNER JOIN lote AS l
    ON pc.codigo_lote = l.codigo
INNER JOIN producto AS p
    ON pc.producto = p.nombre_comercial
INNER JOIN operario_n AS o
    ON pc.documento_operario = o.documento;
```

### 4.3. Consultar los datos normalizados

```sql
SELECT
    a.fecha,
    l.codigo AS codigo_lote,
    l.cultivo,
    p.nombre_comercial AS producto,
    p.categoria_tox,
    a.cantidad,
    o.documento,
    o.nombre_completo,
    o.telefono,
    a.observaciones
FROM aplicacion_n AS a
INNER JOIN lote AS l
    ON a.id_lote = l.id_lote
INNER JOIN producto AS p
    ON a.id_producto = p.id_producto
INNER JOIN operario_n AS o
    ON a.documento_operario = o.documento
ORDER BY a.fecha;
```

## 5. Ejercicio 5. Consultas de detecciones YOLO

### 5.1. Detecciones de plagas por lote y semana

```sql
SELECT
    lote,
    YEARWEEK(fecha_hora, 3) AS anio_semana,
    COUNT(*) AS total_detecciones
FROM v_deteccion_valida
WHERE clase <> 'planta_sana'
GROUP BY
    lote,
    YEARWEEK(fecha_hora, 3)
ORDER BY
    lote,
    anio_semana;
```

### 5.2. Detecciones por hectárea

```sql
SELECT
    lote,
    MAX(area_ha) AS area_ha,
    COUNT(*) AS total_detecciones,
    ROUND(
        COUNT(*) / NULLIF(MAX(area_ha), 0),
        2
    ) AS detecciones_por_ha
FROM v_deteccion_valida
WHERE clase <> 'planta_sana'
GROUP BY lote
ORDER BY detecciones_por_ha DESC;
```

### 5.3. Comparar los umbrales de confianza de la maleza

```sql
SELECT
    0.30 AS umbral_confianza,
    COUNT(*) AS total_detecciones
FROM deteccion
WHERE clase = 'maleza'
  AND confianza >= 0.30

UNION ALL

SELECT
    0.45,
    COUNT(*)
FROM deteccion
WHERE clase = 'maleza'
  AND confianza >= 0.45

UNION ALL

SELECT
    0.60,
    COUNT(*)
FROM deteccion
WHERE clase = 'maleza'
  AND confianza >= 0.60

UNION ALL

SELECT
    0.75,
    COUNT(*)
FROM deteccion
WHERE clase = 'maleza'
  AND confianza >= 0.75;
```

### 5.4. Crear la tabla para comparar versiones del modelo

```sql
CREATE TABLE evaluacion_modelo (
    id_evaluacion INT AUTO_INCREMENT PRIMARY KEY,
    id_imagen INT NOT NULL,
    modelo_version VARCHAR(50) NOT NULL,
    fecha_ejecucion DATETIME NOT NULL
        DEFAULT CURRENT_TIMESTAMP,

    CONSTRAINT uq_imagen_modelo
        UNIQUE (id_imagen, modelo_version),

    CONSTRAINT fk_evaluacion_imagen
        FOREIGN KEY (id_imagen)
        REFERENCES imagen(id_imagen)
        ON DELETE CASCADE
);
```

Agregar la relación a la tabla `deteccion`:

```sql
ALTER TABLE deteccion
ADD COLUMN id_evaluacion INT NULL,
ADD CONSTRAINT fk_deteccion_evaluacion
    FOREIGN KEY (id_evaluacion)
    REFERENCES evaluacion_modelo(id_evaluacion);
```

## 6. Verificaciones finales

### 6.1. Verificar los productos

```sql
SELECT *
FROM producto
ORDER BY nombre_comercial;
```

### 6.2. Verificar las aplicaciones

```sql
SELECT *
FROM aplicacion
ORDER BY fecha;
```

### 6.3. Verificar los operarios

```sql
SELECT *
FROM operario;
```

### 6.4. Verificar la migración

```sql
SELECT COUNT(*) AS total_aplicaciones_normalizadas
FROM aplicacion_n;
```

### 6.5. Verificar las detecciones válidas

```sql
SELECT *
FROM v_deteccion_valida
LIMIT 20;
```

## 7. Notas de ejecución

- Ejecutar las consultas en el orden de los ejercicios.
- Crear y llenar `operario_n` antes de crear y llenar `aplicacion_n`.
- Ejecutar cada instrucción `INSERT` una sola vez, salvo que se hayan eliminado previamente los registros correspondientes.
- Ejecutar la creación de `evaluacion_modelo` y el `ALTER TABLE` una sola vez.
- Verificar que los nombres de las columnas coincidan con los definidos en el cuaderno original.
- Los resultados dependen de los datos que se hayan insertado y de las detecciones que genere YOLO.
````
